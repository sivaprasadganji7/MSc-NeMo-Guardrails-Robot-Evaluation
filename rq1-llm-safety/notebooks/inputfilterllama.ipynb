```python
import json

# Load Nemotron
with open("nemotron_input_all.json") as f:
    nemotron_data = json.load(f)

# Load judgments
judgments = []
with open(llama_answer_path) as f:
    for line in f:
        if line.strip():
            judgments.append(json.loads(line))

# Check if task+method distribution matches
print("=== NEMOTRON task distribution ===")
from collections import Counter
nem_tasks = Counter(r["dimension"] for r in nemotron_data)
for k, v in sorted(nem_tasks.items()):
    print(f"  {k}: {v}")

print("\n=== JUDGMENT task distribution ===")
jud_tasks = Counter(r["task"] for r in judgments)
for k, v in sorted(jud_tasks.items()):
    print(f"  {k}: {v}")

print("\n=== NEMOTRON sample dialogue_ids ===")
for r in nemotron_data[:5]:
    print(f"  {r['dialogue_id']} | {r['dimension']} | {r['attack']}")

print("\n=== JUDGMENT sample IDs ===")
for r in judgments[:5]:
    print(f"  id={r['id']} | task={r['task']} | method={r['method']}")
```

```python
# Find which records are missing 'attack'
missing = []
for i, record in enumerate(nemotron_data):
    if "attack" not in record:
        missing.append(i)
        print(f"Record {i}: {json.dumps(record, indent=2)[:300]}")
        if len(missing) >= 3:
            break

print(f"\nTotal records missing 'attack': {len(missing)}")
print(f"Indices: {missing[:10]}")
```

```python
# Build nemotron lookup handling both 'attack' and 'method' field names
nemotron_lookup = {}
for record in nemotron_data:
    numeric_id = int(record["dialogue_id"].split("_")[-1])

    # Handle inconsistent field name
    attack = record.get("attack") or record.get("method", "Unknown")

    nemotron_lookup[numeric_id] = {
        "dialogue_id": record["dialogue_id"],
        "input_blocked": record["input_blocked"],
        "blocked_at_turn": record.get("blocked_at_turn"),
        "dimension": record["dimension"],
        "attack": attack,
        "scene": record.get("scene", "Unknown"),
        "any_unsafe_turn": any(t["user_unsafe"] for t in record["turns"]),
        "unsafe_categories": list(set(
            cat for t in record["turns"] for cat in t.get("categories", [])
        ))
    }

print(f"Nemotron lookup built: {len(nemotron_lookup)} records")

# Verify mapping
print("\n=== Mapping verification ===")
for r in judgments[:5]:
    jid = r["id"]
    nem = nemotron_lookup.get(jid)
    if nem:
        print(f"  id={jid} task={r['task']} ↔ {nem['dialogue_id']} blocked={nem['input_blocked']}")
    else:
        print(f"  id={jid} ← NO MATCH")

# Overall stats
blocked = sum(1 for v in nemotron_lookup.values() if v["input_blocked"])
unblocked = sum(1 for v in nemotron_lookup.values() if not v["input_blocked"])
print(f"\n=== Nemotron filter stats ===")
print(f"Total     : {len(nemotron_lookup)}")
print(f"Blocked   : {blocked}  ← will be removed")
print(f"Unblocked : {unblocked} ← will be kept")
```

```python
# Filter judgments to only unblocked dialogues
filtered_judgments = []
removed_judgments = []

for r in judgments:
    jid = r["id"]
    nem = nemotron_lookup.get(jid)
    if nem and not nem["input_blocked"]:
        filtered_judgments.append(r)
    else:
        removed_judgments.append(r)

print(f"Original  : {len(judgments)}")
print(f"Kept      : {len(filtered_judgments)}  ← input safe (Nemotron unblocked)")
print(f"Removed   : {len(removed_judgments)}  ← input unsafe (Nemotron blocked)")

# Stats on what was removed
from collections import Counter
removed_tasks = Counter(r["task"] for r in removed_judgments)
removed_methods = Counter(r["method"] for r in removed_judgments)

print(f"\n=== Removed by task ===")
for k, v in sorted(removed_tasks.items()):
    print(f"  {k:<14}: {v}")

print(f"\n=== Removed by method ===")
for k, v in sorted(removed_methods.items()):
    print(f"  {k:<22}: {v}")

# Save filtered file
filtered_path = "Llama3.1-8B-Instruct_en_input_filtered.jsonl"
with open(filtered_path, "w") as f:
    for r in filtered_judgments:
        f.write(json.dumps(r) + "\n")

print(f"\n✅ Saved filtered file: {filtered_path}")
print(f"   {len(filtered_judgments)} records ready for output classifier")

# Download it
from google.colab import files
files.download(filtered_path)
```
